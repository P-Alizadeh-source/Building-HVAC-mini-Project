# 1. Building model

Explore the simple 2R–2C building model and the baseline thermostat.

```
import sys, os
sys.path.append('../src')
import matplotlib.pyplot as plt
from simulation import simulate
```

```
df=simulate(168)
df.head()
```

```
plt.figure(figsize=(10,4)); plt.plot(df.hour,df.T_out,label='Outdoor'); plt.plot(df.hour,df.T_in,label='Indoor'); plt.axhline(20,ls='--'); plt.axhline(24,ls='--'); plt.xlabel('Hour'); plt.ylabel('Temperature [°C]'); plt.legend(); plt.grid(); plt.show()
```

```
print(f"Electrical energy: {df.energy_kWh.sum():.2f} kWh")
print(f"Comfort: {((df.T_in>=20)&(df.T_in<=24)).mean()*100:.1f}%")
```
