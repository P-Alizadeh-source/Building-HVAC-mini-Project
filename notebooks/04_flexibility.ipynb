# 4. Energy flexibility

Use an illustrative time-of-use price profile to compare the electricity cost of the two operating strategies.

```
import sys, os
sys.path.append('../src')
import matplotlib.pyplot as plt
from simulation import simulate
from optimization import simulate_optimized
from flexibility import tou_price, cost
```

```
baseline=simulate(168); optimized=simulate_optimized(168); prices=tou_price(168)
print(f'Thermostat cost: {cost(baseline,prices):.2f}')
print(f'Optimized cost: {cost(optimized,prices):.2f}')
```

```
plt.figure(figsize=(10,4)); plt.plot(prices,label='Illustrative electricity price'); plt.xlabel('Hour'); plt.ylabel('Price [arbitrary units/kWh]'); plt.legend(); plt.grid(); plt.show()
```

```
plt.figure(figsize=(10,4)); plt.plot(baseline.hour,baseline.P_electric/1000,label='Thermostat'); plt.plot(optimized.hour,optimized.P_electric/1000,label='Optimized'); plt.xlabel('Hour'); plt.ylabel('Electrical power [kW]'); plt.legend(); plt.grid(); plt.show()
```
