# 3. HVAC optimisation

Compare a simple thermostat with constrained HVAC optimisation. The optimiser tests a small set of HVAC powers and selects the lowest-energy feasible action for the next hour.

```
import sys, os
sys.path.append('../src')
import matplotlib.pyplot as plt
from simulation import simulate
from optimization import simulate_optimized
```

```
baseline=simulate(168)
optimized=simulate_optimized(168)
for name,d in [('Thermostat',baseline),('Optimized',optimized)]:
    comfort=((d.T_in>=20)&(d.T_in<=24)).mean()*100
    print(name, f'Energy={d.energy_kWh.sum():.2f} kWh', f'Comfort={comfort:.1f}%', f'Peak={d.P_electric.max()/1000:.2f} kW')
```

```
plt.figure(figsize=(10,4)); plt.plot(baseline.hour,baseline.T_in,label='Thermostat'); plt.plot(optimized.hour,optimized.T_in,label='Optimized'); plt.axhline(20,ls='--'); plt.axhline(24,ls='--'); plt.xlabel('Hour'); plt.ylabel('Indoor temperature [°C]'); plt.legend(); plt.grid(); plt.show()
```

```
plt.figure(figsize=(10,4)); plt.plot(baseline.hour,baseline.P_electric/1000,label='Thermostat'); plt.plot(optimized.hour,optimized.P_electric/1000,label='Optimized'); plt.xlabel('Hour'); plt.ylabel('Electrical power [kW]'); plt.legend(); plt.grid(); plt.show()
```
